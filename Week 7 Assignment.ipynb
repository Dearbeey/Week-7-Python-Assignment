```python
# importing Libraries
import pandas
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# style for plots
sns.set(style="whitegrid")
```

```python
# Load the Iris dataset from sklearn
try:
    iris = load_iris(as_frame=True)
    df = iris.frame
    print("First 5 rows of the dataset:")
    display(df.head())

    print("\nDataset Info:")
    display(df.info())

    print("\nMissing values:")
    display(df.isnull().sum())
except Exception as e:
    print("An error occurred while loading the dataset:", e)
```

```python
# Basic Analysis
print("Basic statistics of numerical columns:")
display(df.describe())

# Map target index to species name
df['species'] = df['target'].map(dict(zip(range(3), iris.target_names)))

# Group by species and compute mean
print("\nMean measurements per species:")
display(df.groupby('species').mean())
```

```python
# Line Graph
plt.figure(figsize=(8, 4))
plt.plot(df.index, df['petal length (cm)'], label='Petal Length')
plt.title('Petal Length Trend')
plt.xlabel('Index')
plt.ylabel('Petal Length (cm)')
plt.legend()
plt.tight_layout()
```

```python
# Bar Chart
plt.figure(figsize=(4, 6))
sns.barplot(x='species', y='petal length (cm)', data=df, estimator='mean')
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.tight_layout()
```

```python
# Historygram
import numpy as np
import matplotlib.pyplot as plt

# Data to plot
data = df['sepal width (cm)']

# Histogram data
counts, bins, patches = plt.hist(data, bins=10, edgecolor='black')

# Assign a different color to each bar
colors = plt.cm.viridis(np.linspace(0, 1, len(patches))) 

for patch, color in zip(patches, colors):
    patch.set_facecolor(color)

# Titles and labels
plt.title('Distribution of Sepal Width with Colorful Bars')
plt.xlabel('Sepal Width (cm)')
plt.ylabel('Frequency')
plt.tight_layout()


```

```python
# Scatter Plot
plt.figure(figsize=(6, 4))
sns.scatterplot(x='sepal length (cm)', y='petal length (cm)', hue='species', data=df)
plt.title('Sepal Length vs Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.tight_layout()
```
